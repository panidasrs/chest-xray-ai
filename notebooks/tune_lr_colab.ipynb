## Comparison (baseline = lr 1e-4)

```python
import json
import pandas as pd

runs = {
    "3e-5": "results/resnet18_lr3e-5_test_results.json",
    "1e-4 (baseline)": "results/test_results.json",
    "3e-4": "results/resnet18_lr3e-4_test_results.json",
}
rows = []
for lr, path in runs.items():
    r = json.load(open(path))
    hist = pd.read_csv(path.replace("test_results.json", "history.csv"))
    rows.append({"lr": lr, "best_val_auc": round(hist["val_auc"].max(), 4), **r})

table = pd.DataFrame(rows)[["lr", "best_val_auc", "test_auc", "sensitivity", "specificity", "accuracy", "fn", "fp", "threshold"]]
table.to_csv("results/lr_tuning.csv", index=False)
table
```
